```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("employee_data.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
#Calculate Attrition Rate

attrition_rate = (
    df["Attrition"].value_counts()["Yes"]
    /
    len(df)
)*100

print(attrition_rate)
```

```python
#Analyze Attrition by Department

department_attrition = (
df.groupby("Department")["Attrition"]
.value_counts()
)

print(department_attrition)
```

```python
#Visualize Department Attrition
sns.countplot(
    x = "Department",
    hue="Attrition",
    data=df
)

plt.titile(
    "Employee Attrition by Department"
)

plt.show()
```

```python
#Analyze Overtime Impact
overtime_analysis = (
    df.groupby("Overtime")["Attrition"].value_counts()
)

print(overtime_analysis)
```

```python
#Salary Analysis

sns.boxplot(
    x="Attrition",
    y="Monthly_Income",
    data=df
)

plt.show()
```
